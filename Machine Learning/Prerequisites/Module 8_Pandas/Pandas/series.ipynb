## Pandas Series is a column of data in a table like in excel or google sheets with labels called as indexes attached to each value.

```python
pd.Series(data=[40,50,60], index=['Bread','Milk','Butter'])
```

```python
pd.Series(data=[40,50,60], index=['Bread','Milk','Butter'], name='Price of grocery items')
```

#### Series is a 1-d array that can hold any type of data like integers, floats, boolean, string (objects), mixed data types

## Data type ---> integers --> data in a Series (column) --> Quantity of items in the stock
## Data type ---> Float ---> data in a Series (column) ---> Price or product ratings
## Data type ---> String --> data in a Series (column) ---> Product name
## Data type ---> boolean --> data in a Series (column) ---> Is item in stock ?
## Data type ---> Data/ Time --> data in a Series (column) ---> Expiry dates or sale dates
## Mixed type ---> Milk, 40, True

## Series methods

#### 1) head() ---> provides the top preview of data as the top 5 rows of the dataframe or Series. It is quite helpful with providing the preview of the dataset and if we want to preview top 10 rows then we can pass 10 as the argument or for top 7 rows we can pass 7 as an argument. 

#### 2) tail() ---> provides the bottom preview of data as the bottom 5 rows

#### for e.g. --> dataset --> salary column --> top 5 rows --> salary --> 7 lpa ---> top 10 ---> top 15 ,,, tail() ---> bottom 5 rows ---> salary --> 4 lpa, 9 lpa

#### 3) sample() --> provides any custom row from the dataset for us to preview it.

#### 4) value_counts() --> it provides the idea about the count of the values (categorical values), if i am having a column where the categorical value count i want to calculate then i can go with value_counts()

#### 5) sort_values() --> this method sorts the value in an ascending manner. It can be useful for us for the numerical columns. you can even sort the values in descending manner as well.

#### 6) sort_index() --> this method sorts the index of the dataset.

### Mathematical functions in Pandas

#### 1) sum() --> sum of all values present inside the series
#### 2) product() --> product of all values in the series
#### 3) mean() --> provides us with the average value of the data column.
#### for e.g. --> company data --> salary --> average value for that column --> salary = 6,6,8,9,11,,, mean = (6+6+8+9+11)/5 --> 8 LPA

#### median() --> also provides the idea about the average of the data. for coming up with median, we need to sort the values in ascending manner and then the middle value after sorting the data --> median; data ----> salary --> 70 lakh (outlier), ---> avg --> 7+8+6+9+ 

#### mode() --> mode() --> provides us with the idea that which value is getting repeated maximum number of times.

#### std() --> if i need to get thet idea that how the values inside the series are distributed from the average or mean value then we can come up with the standard deviation
#### salary --> 7,8,6,6.5,7.5 --> low standard deviation
#### salary --> 7,15,4,20,28,17 --> high standard deviation

#### variance --> variance also provides the insight about the distribution of the data from the mean value and variance is the square of the standard deviation.
#### standard deviation = (|mean - variance 1| + |mean - variance 2| + ...) / total count of values

#### variance = std**2

#### min() --> minimum value in the series
#### max() --> maximum value in the series


## describe () --> provides the idea about the summary of the important mathematical calculations for the data.

### below ideas we'll be getting after using describe function
#### min --> minimum value
#### max --> maximum value
#### mean --> average value
#### std --> standard deviation
#### 25%ile --> 25% of the values out of my all values are less than this specific value
#### 50% --> 50% of the values out of my all values are less than this specific value
#### 75% --> 75% of the values out of my all values are less than this specific value
#### count ----> total number of rows in the column

#### JEE Mains --> 25% --> my score in JEE Mains was greater than 25% of the total applicants who applied in JEE
