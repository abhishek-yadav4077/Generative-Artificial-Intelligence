## Pandas Series is a column of data in a table like in excel or google sheets with labels called as indexes attached to each value.

```python
pd.Series(data=[40,50,60], index=['Bread','Milk','Butter'])
```

```python
pd.Series(data=[40,50,60], index=['Bread','Milk','Butter'], name='Price of grocery items')
```

#### Series is a 1-d array that can hold any type of data like integers, floats, boolean, string (objects), mixed data types

## Data type ---> integers --> data in a Series (column) --> Quantity of items in the stock
## Data type ---> Float ---> data in a Series (column) ---> Price or product ratings
## Data type ---> String --> data in a Series (column) ---> Product name
## Data type ---> boolean --> data in a Series (column) ---> Is item in stock ?
## Data type ---> Data/ Time --> data in a Series (column) ---> Expiry dates or sale dates
## Mixed type ---> Milk, 40, True 

## Creating a Series in Multiple ways :-
#### 1. List
#### 2. NumPy array
#### 3. Dictionary
#### 4. Custom index
#### 5. Scalar value

```python
## Series from list:-

country=['India','China','USA','Pakistan']

pd.Series(country)
```

```python
batsman_runs = [48,64,72,98,114,183]


pd.Series(batsman_runs)
# pd.Series(batsman_runs, dtype='int32')
```

```python
subject_names = ['Maths','Physics','Chemistry','Biology','English']

subject_marks = [48,92,64,35,98]

pd.Series(data=subject_marks, index=subject_names, name='Subject marks data', dtype='int32')
```

```python
### Creating a Series using a NumPy array

values = np.arange(10,21)
pd.Series(values)
```

```python
## Dictionary to Series

## Dictionary is another data structure that stores the data in a key:value pair and 
## just because it stores the data in a key:value pair so we don't have to explicitly mention the index.
```

```python
data_items = {'Marks':78, 'English':92, 'Physics':88, 'Chemistry':64}
pd.Series(data_items)
```

```python
## custom index to series

values = [1,2,3,4,5]
pd.Series(data=values, index=['a','b','c','d','e'])
```

```python
## Scalar value inside a Series 

pd.Series(4,index=['a','b','c','d','e'])
```

## Series methods

#### 1) head() ---> provides the top preview of data as the top 5 rows of the dataframe or Series. It is quite helpful with providing the preview of the dataset and if we want to preview top 10 rows then we can pass 10 as the argument or for top 7 rows we can pass 7 as an argument. 

#### 2) tail() ---> provides the bottom preview of data as the bottom 5 rows

#### for e.g. --> dataset --> salary column --> top 5 rows --> salary --> 7 lpa ---> top 10 ---> top 15 ,,, tail() ---> bottom 5 rows ---> salary --> 4 lpa, 9 lpa

#### 3) sample() --> provides any custom row from the dataset for us to preview it.

#### 4) value_counts() --> it provides the idea about the count of the values (categorical values), if i am having a column where the categorical value count i want to calculate then i can go with value_counts()

#### 5) sort_values() --> this method sorts the value in an ascending manner. It can be useful for us for the numerical columns. you can even sort the values in descending manner as well.

#### 6) sort_index() --> this method sorts the index of the dataset.
