### Data exploration --> Attributes # Functions for the dataFrame

### Attributes

#### 1. shape --> it provides the info about the (row, column) information about the dataset
#### syntax --> df.shape

#### 2. dtype --> It provides the idea about the datatype of all columns, so all columns with their name and their associated datatype will be shown by using dtypes
#### syntax --> df.dtypes

#### 3. index --> for a dataframe with labelled index we can get the idea about the labelled just by using index attributes

#### 4. columns --> it provides the idea about all of the column names of the given dataframe

#### 5. values --> it provides the idea about the values present row by row inside the dataframe. It always returns a 2d numpy array representing the values of column row by row

#### 6. ndim --> it provides the idea about the dimensions of the dataframe

#### 7. size --> total number of elements present inside the dataframe, that means all rows * all columns,,, (column 1 --> 50 rows, column - 2 ---> 50 rowsm, column 3 --> 70 rows)----> size ---> 50 + 50 + 70

## Functions

#### 1. head() --> to get to have the idea about the data that your dataset is storing as the top 5 rows, that means that starting from the top we can get the top 5 rows and along with that if i want that instead of top 5 rows i want to see a fixed number of rows from the top , that number i can pass inside the head() as a parameter 

#### 2. tail() --> provide the insight about the dataframe from the bottom so tail() can show us the bottom 5 rows and again based on the values you will be passing as the parameter, those many values you can get using the tail()

#### 3. sample() --> provides the insight about the dataframe as the random values present inside the dataset and again based on how many values i want i can put the parameter as per that.

#### 4. info() --> provides the idea ---> 
#### i. about the datatype of all columns
#### ii. number of missing values of all columns
#### iii. dataframe storage in memory

#### 5. describe() --> it provides an idea as the mathematical summary of the data --> minimum, maximum, count, mean, std, 25%, 50%, 75%
