# NUMPY

#### NumPy is a fundamental package for scientific computing in python. It is a python library that provides a multi dimentional array objects and various derived objects such as masked arrays, matrices and an assortment of routines for fast operations on arrays including mathematical, logical, shape manipulation, sorting, selecting, i/o, discrete fourier transform, basic linear algebra, statistical operations, random simulations etc.

#### At the core of the NumPy package is the n-dimentional array objects. 

## Why NumPy ?

### 2008-2009, Data science was boom --> Matlab, R language (were popular not python)

## Data types in python --> list, dictionaries --> slow
Syntax --> Python --> Easy
## We needed speed and less memory to work with the huge amount of data.

NumPy --> C language --> wrapped --> python language
## NumPy --> syntax --> easy for beginners --> performing scientific calculations --> creating multi dimentional arrays --> speed is faster for applying operations --> less memory

### Libraries like scikit learn, tensorflow and pytorch requires input data in NumPy arrays or tensors.

## NumPy arrays vs python sequences

#### NumPy arrays have a fixed size at creation, unlike python lists which grow dynamically. Changing the size of an n-dimentional array will create a new array and delete the original.

#### The elements in a NumPy array are required to be of the same data type and thus will be of the same size in memory.

#### NumPy arrays facilitates advanced mathematical and other types of operations on large number of data. Typically, such operations are executed more efficiently with less code than is possible using python's built in sequences.

## Stacking and Splitting in arrays --> size of the array must be the same

## Stacking --> joining the array together
## Horizontal stacking --> row by row joining is done --> hstack()
## Vertical stacking --> column by column joining is done --> vstack()




```python
array_one = np.arange(1,13).reshape(3,4)
```

```python
array_one
```

```python
array_two = np.arange(13,25).reshape(3,4)
```

```python
array_two
```

```python
horizontal_array = np.hstack((array_one, array_two))
```

```python
horizontal_array
```

```python
horizontal_array = np.hstack((array_one, array_two, array_one, array_two))
```

```python
horizontal_array
```

```python
vertical_array = np.vstack((array_one, array_two))
```

```python
vertical_array
```

## Horizontal split --> np.hsplit(array, number of parts you want your array to be splitted at)
## Vertical splits --> np.vsplit(array, number of parts you want your array to be splitted at)

```python
np.hsplit(vertical_array, 2)
```

```python
np.vsplit(vertical_array, 2)
```

```python
np.vsplit(vertical_array, 3)
```
